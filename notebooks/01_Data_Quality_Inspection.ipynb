# E-Commerce Sales & Customer Analytics

## 01 — Data Quality Inspection

This notebook performs the initial inspection of the Online Retail II dataset.

### Objectives
- Load the raw transaction data
- Understand the dataset structure
- Inspect data types
- Identify missing values
- Identify duplicates
- Investigate cancellations and returns
- Check invalid quantities and prices
- Establish the date range
- Prepare the dataset for cleaning

```python
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
```

```python
file_path = "../data/online_retail_II.xlsx"

print("Dataset path:", file_path)
```

```python
df_2009 = pd.read_excel(
    file_path,
    sheet_name="Year 2009-2010"
)

print("Rows:", f"{len(df_2009):,}")
print("Columns:", df_2009.shape[1])
```

```python
df_2009.head()
```

```python
df_2009.info()
```

```python
missing = pd.DataFrame({
    "Missing Values": df_2009.isnull().sum(),
    "Missing %": (df_2009.isnull().mean() * 100).round(2)
})

missing
```

```python
duplicate_count = df_2009.duplicated().sum()

print("Duplicate rows:", f"{duplicate_count:,}")
```

```python
cancelled = df_2009["Invoice"].astype(str).str.startswith("C")

print("Cancelled transactions:", f"{cancelled.sum():,}")
print("Cancelled %:", f"{cancelled.mean() * 100:.2f}%")
```

```python
print("Negative quantities:", (df_2009["Quantity"] < 0).sum())
print("Zero quantities:", (df_2009["Quantity"] == 0).sum())
print("Positive quantities:", (df_2009["Quantity"] > 0).sum())
```

```python
print("Negative prices:", (df_2009["Price"] < 0).sum())
print("Zero prices:", (df_2009["Price"] == 0).sum())
print("Positive prices:", (df_2009["Price"] > 0).sum())
```

```python
print("First transaction:", df_2009["InvoiceDate"].min())
print("Last transaction:", df_2009["InvoiceDate"].max())
```

```python
df_2009["Revenue"] = df_2009["Quantity"] * df_2009["Price"]

print("Total transaction value:", f"£{df_2009['Revenue'].sum():,.2f}")
```

```python
df_2009["Revenue"].describe()
```

```python
missing_2010 = pd.DataFrame({
    "Missing Values": df_2010.isnull().sum(),
    "Missing %": (df_2010.isnull().mean() * 100).round(2)
})

missing_2010
```

```python
print("Negative quantities:", (df_2010["Quantity"] < 0).sum())
print("Zero quantities:", (df_2010["Quantity"] == 0).sum())
print("Positive quantities:", (df_2010["Quantity"] > 0).sum())

print()

print("Negative prices:", (df_2010["Price"] < 0).sum())
print("Zero prices:", (df_2010["Price"] == 0).sum())
print("Positive prices:", (df_2010["Price"] > 0).sum())
```
